```python
for i in range(10):
    print(i + 1)
```

```python
# Recorrer una lista de nombres y saludar a cada uno.
personas = ["Juanito", "Pedrito", "Camilito", "Jerick", "Fernando"]

for i in personas:
    print(f"Hola {i} como estas?")
```

```python
# Usar un bucle while para contar hasta que el usuario escriba "salir".
respuesta = ""
numero = 0
while respuesta != "exit":
    respuesta = input("Para salir escribe exit")
    numero += 1
    print(numero)
```

```python
# Generar la tabla de multiplicar de un número ingresado
# por el usuario utilizando un bucle for.
numero = int(input("Ingresa un numero para la tabla de multiplicar"))
for i in range(1,11):
    print(f"{numero} x {i} = {i * numero}")
```
